## Veh Displacement y descubrimiento de valores raros para electricos (cleaning tambien)

Al investigar los vehiculos con displacement con ese valor raro, descubrimos que todos los que tenian ese valor raro eran electricos. Luego de eso, nos fijamos segun el tipo de combustible que eran electricos, como se distrbuian los valores de Veh Displacement para ellos, y ahi vimos que no tenian valores nulos (cuando deberian tenerlos) y los limpiamos para que tengan valores iguales a pd.NA

```python
import numpy as np
values_to_replace = [0.001, 99.999, 0.010, 99.990, 1.000]

condicion = (epa_cartest_clean["Test Fuel Type Cd"] == 62) & (epa_cartest_clean["Test Veh Displacement (L)"].isin(values_to_replace))

epa_cartest_clean.loc[condicion, "Test Veh Displacement (L)"] = np.nan
```
